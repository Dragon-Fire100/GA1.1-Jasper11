```python
def subtract(a,b,):
    D = a - b
    return D

print(subtract(1,3))
print(subtract(8,5))
print(subtract(150,100))
```

```python
def distance_from_zero(x):
    if x >= 0:
        return subtract(x, 0)
    else:
        return subtract(0, x)

print(distance_from_zero(5))
print(distance_from_zero(-3))
print(distance_from_zero(2))
```
