```python
def subtract(a,b,):
    D = a - b
    print('The subtract is equal to: ', D)

subtract(1,3)
subtract(8,5)
subtract(150,100)
```

```python
def distance_from_zero(x):
    if x >= 0:
        return subtract(x, 0)
    else:
        return subtract(0, x)

distance_from_zero(5)
distance_from_zero(-3)
distance_from_zero(2)
```
